## Mathematical Model of a Sine Wave

A continuous sinusoidal wave can be represented mathematically as:

$$s(t) = A \cdot \sin(4 \pi f t + \phi)$$

Where:
- $t$: Time variable (seconds)
- $A$: Amplitude (peak value of the signal)
- $f$: Frequency (Hz, number of cycles per second)
- $\phi$: Initial phase angle (radians)

For this initial draft, we set $A = 1$, $f = 1\text{ Hz}$, and $\phi = 0$, simplifying the equation to:

$$s(t) = \sin(4 \pi t)$$

```python
import matplotlib.pyplot as plt
import numpy as np

# Create time array: 1000 values from 0 to 1
t = np.linspace(0, 1, 1000)

# Create sinusoidal signal array
s = np.sin( 4 * np.pi * t )

# Plot s as a function of t
plt.figure(figsize=(8, 4))
plt.plot(t, s, label="s(t) = sin(4πt)")

# Plot formatting
plt.title("Sinusoidal Signal Draft Plot")
plt.xlabel("Time t (s)")
plt.ylabel("Signal Amplitude s(t)")
plt.grid(True)
plt.legend()

# Save plot as PNG file
plt.savefig("sinusoid_draft.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# 1. Create a figure and the primary axis (ax1)
fig, ax1 = plt.subplots(figsize=(8, 4))

# 2. Plot your original Sine wave on the primary (left) y-axis
# (We reuse your 't' and 's' from the previous cell)
ax1.plot(t, s, color='blue', label="s(t) = sin(4πt)")
ax1.set_xlabel("Time t (s)")
ax1.set_ylabel("Sine Amplitude s(t)", color='blue') # First y-axis label
ax1.tick_params(axis='y', labelcolor='blue')
ax1.grid(True)

# 3. Create a secondary (right) y-axis sharing the same x-axis
ax2 = ax1.twinx()

# 4. Calculate and plot Cosine function on the secondary y-axis
# (Matching your frequency: 4 * np.pi * t)
c = np.cos(4 * np.pi * t)
ax2.plot(t, c, color='red', linestyle='--', label="c(t) = cos(4πt)")
ax2.set_ylabel("Cosine Amplitude c(t)", color='red') # Second y-axis label
ax2.tick_params(axis='y', labelcolor='red')

# 5. Add a legend that combines both plots (Sine and Cosine)
# Extract handles and labels from both axes to put them in a single legend
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
ax1.legend(lines_1 + lines_2, labels_1 + labels_2, loc="upper right")

# Title and Save as PDF
plt.title("Sine vs Cosine Signal with Dual Y-Axis")
plt.savefig("sinusoid_better.pdf", dpi=300, bbox_inches="tight")

# Show the plot
plt.show()
```

```python
def sinusoid(time_array, amplitude, frequency, phase):
    """
    Generates a sinusoidal signal array based on physical parameters.

    Parameters:
    -----------
    time_array : numpy.ndarray
        1D array representing the time vector (in seconds).
    amplitude : float
        The peak amplitude of the sinusoid.
    frequency : float
        The frequency of the wave (in Hertz).
    phase : float
        The phase shift (in radians).

    Returns:
    --------
    x : numpy.ndarray
        The generated signal amplitude at each time point.
    """
    # We multiply by 2*pi to convert frequency (Hz) to angular frequency (rad/s)
    # This ensures the formula mathematically aligns with y(t) = A * sin(wt + phi)
    angular_frequency = 2 * np.pi * frequency
    
    # Store the result in a variable named 'x' as explicitly requested in the assignment
    x = amplitude * np.sin(angular_frequency * time_array + phase)
    
    return x

# Generate two signals with different frequencies to visually compare their cycles.
# We reuse the global time array 't' created in the first draft.
s1 = sinusoid(time_array=t, amplitude=1.0, frequency=2.0, phase=0.0)
s2 = sinusoid(time_array=t, amplitude=1.0, frequency=5.0, phase=0.0)

# Plot s1 and s2 on the same figure to observe the frequency difference
plt.figure(figsize=(8, 4))
plt.plot(t, s1, color='blue', label='s1: 2 Hz')
plt.plot(t, s2, color='green', label='s2: 5 Hz')

plt.title("Comparison of Sinusoids with Different Frequencies")
plt.xlabel("Time t (s)")
plt.ylabel("Signal Amplitude")
plt.grid(True)
plt.legend()

# Save the final plot
plt.savefig("sinusoids_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```
